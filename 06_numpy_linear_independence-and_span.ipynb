# Rank of a Matrix and Linear Independence

In linear algebra, the **rank of a matrix** represents the number of
linearly independent columns (or rows).

- If columns are linearly dependent → rank < number of columns  
- If columns are linearly independent → rank = number of columns  

Geometric meaning:
- Rank = number of independent directions
- Rank = dimension of the span of the column vectors


```python
import numpy as np
```

## Example 1: Linearly Dependent Vectors

Here, `v2` is equal to `2 × v1`, meaning it does not add a new direction.

So we expect the rank to be less than the number of vectors.


```python
v1 =  np.array([1, 2, 3])
v2 =  np.array([2, 4, 6]) # 2 * v1 (linearly dependent)
v3 =  np.array([1, 0, 1])
```

```python
A =  np.column_stack([v1, v2, v3])  # Create a matrix A by stacking the vectors as columns
print("Matrix A:")
```

```python
rank = np.linalg.matrix_rank(A)  # Calculate the rank of the matrix A
print("Rank of A:", rank)
```

### Interpretation

There are 3 vectors, but the rank is 2.

This means:
- Only 2 independent directions exist.
- The vectors span a 2D plane inside ℝ³.
- Therefore, the vectors are linearly dependent.


## Example 2: Linearly Independent Vectors (Standard Basis)

The standard basis vectors in ℝ³:

e₁ = [1, 0, 0]  
e₂ = [0, 1, 0]  
e₃ = [0, 0, 1]

Each vector represents a unique direction in 3D space.


```python
v4 =  np.array([1,  0, 0])
v5 =  np.array([0, 1, 0])
v6 =  np.array([0, 0, 1])
```

```python
B =  np.column_stack([v4, v5, v6])  # Create a matrix B by stacking the standard basis vectors as columns
print("Matrix B:")
rank_B = np.linalg.matrix_rank(B)  # Calculate the rank of the matrix B
print("Rank of B:", rank_B)
```

### Interpretation

Rank is 3.

This means:
- All 3 vectors are independent.
- They span the entire ℝ³ space.
- Dimension of the span = 3.


# Matrix as a Linear Transformation

A matrix can be viewed as a function that transforms a vector.

Given:

y = D @ x

This represents applying a linear transformation D to vector x.


```python
D = np.array([[1,2],
              [3,4]])
```

```python
x = np.array([1,1])
```

```python
y = D @ x
```

```python
print("Result of D @ x:", y)
```

### Interpretation

The matrix D transforms vector x into a new vector y.

Matrix multiplication is not just arithmetic —
it represents:

- Rotation
- Scaling
- Shearing
- Reflection

In robotics, matrix multiplication is used to:
- Transform coordinate frames
- Rotate vectors
- Compute velocities
- Apply robot joint motions


# Key Concepts Learned

- Rank = number of independent directions.
- Dimension of span = rank.
- A matrix represents a linear transformation.
- Matrix multiplication applies that transformation.
- Independence determines whether vectors add new motion directions.

These concepts are foundational for:
- Jacobians
- Singularities
- Kinematics
- SE(3) transformations
- Robot motion
